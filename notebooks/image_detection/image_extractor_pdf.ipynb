```python
import fitz
import io
from PIL import Image

def extract_image_pages(pdf_path):
    doc = fitz.open(pdf_path)
    image_pages = []

    for page_num in range(len(doc)):
        page = doc.load_page(page_num)
        images = page.get_images(full=True)
        if images:
            image_pages.append(page_num)
    return image_pages
```

```python
def get_page_image(pdf_path, page_num):
    doc = fitz.open(pdf_path)
    page = doc.load_page(page_num)
    pix = page.get_pixmap(dpi=300)
    img_data = io.BytesIO(pix.tobytes())
    image = Image.open(img_data)
    return image
```

```python
import fitz
from pathlib import Path

ROOT_DIR = Path("/mnt/tank/scratch/plutskyi/papers")
OUT_ROOT = Path("/mnt/tank/scratch/plutskyi/cvat_share_files/SEM_TEM_DETECT")
OUT_ROOT.mkdir(parents=True, exist_ok=True)

BATCH_SIZE = 20

def chunked(iterable, n):
    iterable = list(iterable)
    for i in range(0, len(iterable), n):
        yield iterable[i:i + n]

def process_batch(batch_pdfs, task_idx):
    task_dir = OUT_ROOT / f"task_{task_idx}"
    task_dir.mkdir(parents=True, exist_ok=True)
    
    images_count = 0
    
    for pdf_path in batch_pdfs:
        doc = fitz.open(pdf_path)
        pdf_stem = pdf_path.stem.replace("@", "")

        for i in range(doc.page_count):
            page = doc.load_page(i)
            pix = page.get_pixmap(dpi=300)
            out_path = task_dir / f"{pdf_stem}_page_{i+1:04d}.png"
            pix.save(out_path)
            images_count += 1

        doc.close()
    return images_count

def work():
    pdfs = sorted(list(ROOT_DIR.rglob("*.pdf")))
    total_images = 0
    for task_idx, batch in enumerate(chunked(pdfs, BATCH_SIZE), start=1):
        images_count = process_batch(batch, task_idx)
        total_images += images_count
        print(f"Таска {task_idx}/47 создана, в таске лежит {images_count} png изображения/ий перехожу к следующей таске - {task_idx + 1}")
    
    print(f"Конец работы скрипта, всего изображений: {total_images}")
```

```python
all_chunks = list(chunked(sorted(list(ROOT_DIR.rglob("*.pdf"))), 20))
print("Количество тасков которое мы получим:", len(all_chunks))
print(all_chunks)
```
